```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import make_scorer, matthews_corrcoef
from sklearn.datasets import make_classification
```

```python
mcc_scorer = make_scorer(matthews_corrcoef)
```

```python
def evaluacion_individual(mask, X, y):
  n_features = int(mask.sum())

  if n_features == 0 or n_features> 4:
    return -1.0

  X_subset = X[:, mask.astype(bool)]

  model = LogisticRegression(
      solver="lbfgs",
      max_iter=5000
  )

  cv = StratifiedKFold(
      n_splits=5,
      shuffle=True,
      random_state=42
  )

  scores = cross_val_score(
      model,
      X_subset,
      y,
      cv=cv,
      scoring=mcc_scorer
  )

  return scores.mean()
```

```python
def random_individual(n_features=10, max_features=4):
  while True:
    mask = np.random.randint(
        0,2,size=n_features
    )

    if 1<=mask.sum()<=max_features:
      return mask
```

```python
def two_point_crossover(a,b):
  points = sorted(
      np.random.choice(
          np.arange(1, len(a)),
          size = 2,
          replace=False
    )
  )
  p1, p2 = points
  child1=a.copy()
  child2=b.copy()

  child1[p1:p2], child2[p1:p2] = (
      a[p1:p2],
      b[p1:p2]
  )

  return child1, child2
```

```python
def bit_flip_mutation(individual, indpb=0.2):
  child = individual.copy()
  for i in range(len(child)):
    if np.random.rand() < indpb:
      child[i] = 1 - child[i]

    return child
```

```python
np.random.seed(1)

population_size = 30
n_generations = 12

crossover_probability = 0.5
mutation_probability = 0.2

population = [
    random_individual()
    for _ in range(population_size)
]

history = []

for generation in range(n_generations + 1):

    # 1. Evaluar población
    fitnesses = np.array([
        evaluacion_individual(ind, X, y)
        for ind in population
    ])

    # 2. Identificar mejor individuo
    best_index = np.argmax(fitnesses)

    best_individual = (
        population[best_index]
    )

    best_fitness = (
        fitnesses[best_index]
    )

    mean_fitness = (
        fitnesses.mean()
    )

    history.append({
        "generation": generation,
        "best_MCC": best_fitness,
        "mean_MCC": mean_fitness
    })

    print(
        f"Generation {generation:02d} | "
        f"Best MCC = {best_fitness:.3f} | "
        f"Mean MCC = {mean_fitness:.3f}"
    )

    # No crear nueva población después
    # de la última generación
    if generation == n_generations:
        break

    # 3. Elitismo:
    # conservar el mejor
    new_population = [
        best_individual.copy()
    ]

    # 4. Generar descendientes
    while len(new_population) < population_size:

        parent1 = torneo(
            population,
            fitnesses
        )

        parent2 = torneo(
            population,
            fitnesses
        )

        child1 = parent1.copy()
        child2 = parent2.copy()

        # Crossover
        if np.random.rand() < crossover_probability:

            child1, child2 = (
                two_point_crossover(
                    parent1,
                    parent2
                )
            )

        # Mutation
        if np.random.rand() < mutation_probability:
            child1 = bit_flip_mutation(
                child1
            )

        if np.random.rand() < mutation_probability:
            child2 = bit_flip_mutation(
                child2
            )

        # Reparar si excede máximo
        if child1.sum() > 4:
            ones = np.where(child1 == 1)[0]

            while child1.sum() > 4:
                child1[
                    np.random.choice(ones)
                ] = 0
                ones = np.where(child1 == 1)[0]

        if child1.sum() == 0:
            child1[
                np.random.randint(10)
            ] = 1

        if child2.sum() > 4:
            ones = np.where(child2 == 1)[0]

            while child2.sum() > 4:
                child2[
                    np.random.choice(ones)
                ] = 0
                ones = np.where(child2 == 1)[0]

        if child2.sum() == 0:
            child2[
                np.random.randint(10)
            ] = 1

        new_population.append(child1)

        if len(new_population) < population_size:
            new_population.append(child2)

    population = new_population
```
